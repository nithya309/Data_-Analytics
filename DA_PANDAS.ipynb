# Copy() method

```python
# Def: copy() method is used to create a new independent a new independent copy/object,
#             avoid the accidental change in original data.

# Type of copy operations:
# 1. df1 = df : Assignment : Giving two names to same dataframe
# 2. deep copy : New duplicate copy of dataset
# 3. Shallow copy :New dataset structure but common shared data
```

```python
# Example - 1 : Here df1 & df are just two different name for same data
df1 = df
df1
```

```python
# Example-2 : Deep copy : This will create a separate copy

fav_places = pd.DataFrame({
                     'Place' : ['ladakh', 'Kerala', 'Gachiboli', 'Kokapat'],
                      'Package': [ 80,60,10,15]
})

new_copy = fav_places.copy()
new_copy.iloc[0,0] = 'Dubai'
fav_places
```

```python
# Example-3 : Shallow copy : deep = False

#New dataset structure but common shared data 
fav_places = pd.DataFrame({
                     'Place' : ['ladakh', 'Kerala', 'Gachiboli', 'Kokapat'],
                      'Package': [ 80,60,10,15]
})

new_copy = fav_places.copy(deep = True)
new_copy.iloc[0,0] = 'NewYork'
fav_places
```

# Merging two dataframes

```python
# merge() function can be used to combine two dataframes based upon common column
# merge() functiom works same as SQL join operation : inner,outer,left,right

df1 = pd.DataFrame({'ID' :[1,2,3],'Dream_company': [ 'Amazon','Delloite','Google']})
df2 = pd.DataFrame({'ID' :[2,3,4],'Dream_Package' : [25,30,40,]})

# showimg the inner merge : combine only common records
merged_df = pd.merge(df1,df2, on ='ID', how = 'inner')
merged_df
```

```python
merged_df = pd.merge(df1,df2, on ='ID', how = 'outer')
merged_df
```

```python
merged_df = pd.merge(df1,df2, on ='ID', how = 'left')
merged_df
```

```python
merged_df = pd.merge(df1,df2, on ='ID', how = 'right')
merged_df
```

```python
# select the randaom date/time and build the dataframe
import numpy as np

#choice() function is helping to select the random 1000 data values
random_dates = np.random.choice(time_series_data, size = 1000)
len(random_dates)
```

```python
# create dataframe with date/time column

#pd.to_datatime() : used to covert numpy date/time data into Pandas date/time format
#np.arange(1,1001) : used to generate values from 1 to 1000
np.random.seed(42) # this will help to generate same random values
df = pd.DataFrame({
                    'timestamp' : pd.to_datetime(random_dates),
                    'order_id' : np.arange(1,1001),
                    'food_items' : np.random.choice(['Biryani', 'Egg Puff', 'Paratha', 'Dosa', 'SambharRice'], size = 1000)
})

df['year'] = df['timestamp'].dt.year
df['month'] = df['timestamp'].dt.month
df['day'] = df['timestamp'].dt.day
df['minute'] = df['timestamp'].dt.minute
df['second'] = df['timestamp'].dt.second


df
```
