# 8.Concatenation, stacking & splitting of numpy arrays

```python
# Concatenation() : this function will put content of two or more numpy array into a single array

arr1 = ['Kajal','Anushwara','Rukmini']
arr2 = ['Prabhas', 'Mahesh babu', 'Nani']

new_arr = np.concatenate((arr1,arr2))


print(new_arr)
```

```python
# Query : Concatenate 3 arrays
arr1 = ['Kajal','Anushwara','Rukmini']
arr2 = ['Prabhas', 'Mahesh babu', 'Nani']
arr3 = ['Bramanandam', 'Satya', 'Sunil']

new_arr = np.concatenate((arr1,arr2,arr3))


print(new_arr)
```

```python
# Concatenate 2D arrays : Horizontal concatenation (column wise / side by side)
# axis = 1 : Horizontal concatenation
Dream_package_boys = np.array([[50,10,100],
                          [200,500,300]])
Dream_package_girls = np.array([[12,10,80],
                                [700,50,8000]])
new_2d_arr = np.concatenate((Dream_package_boys,Dream_package_girls), axis = 1)
new_2d_arr
```

```python
# Concatenate 2D arrays : vertical concatenation (row wise)
# axis =  : vertical concatenation
Dream_package_boys = np.array([[50,10,100],
                          [200,500,300]])
Dream_package_girls = np.array([[12,10,80],
                                [700,50,8000]])
new_2d_arr = np.concatenate((Dream_package_boys,Dream_package_girls), axis = 0)
new_2d_arr
```

```python
# stack() : stack function will merge arrays with new dimension

arr1 = np.array([[10,20,30],
                [10,20,30]])
arr2 = np.array([[40,50,60],
                [40,50,60]])

stacked_arr = np.stack((arr1,arr2), axis = 0)
concat_arr = np.concatenate((arr1,arr2), axis = 0)
print(stacked_arr)
print(concat_arr)
```

```python
# column wise stacking
stacked_arr = np.stack((arr1,arr2), axis = 1)
stacked_arr
```
