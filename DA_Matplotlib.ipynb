# Bar chart

```python
# USE: Bar chart is used to represent Categorical and Discrete data.


# Categorical data : e.g Gender, Blood group, True/False
# Continuous data : which is in decimal, e.g Age, Temperature, Height
# Discrete data : Numerical data without decimal value. e.g  no.of students


breakfast_items = ['Idli','Dosa','vada','Puri','Bonda','Pongal']
No_plates = [2,4,20,10,1,5]


plt.figure(figsize = (4,2))  # plt.figure for adjusting the size of the chart
plt.bar(breakfast_items,No_plates)  # plt.bar(x,y)
```

```python
plt.figure(figsize = (4,2))  # plt.figure for adjusting the size of the chart
plt.bar(breakfast_items,No_plates, color = "violet")  # plt.bar(x,y)
```

```python
plt.figure(figsize = (4,2))  # plt.figure for adjusting the size of the chart
plt.bar(breakfast_items,No_plates, color = ["violet", "green", "pink", "orange","blue", "gold"])  # plt.bar(x,y)
```

```python
# Drawback of bar chart: Bar chart cannot handle big item names

breakfast_items = ['Idli','Dosa','vada','Puri','Bonda','Pongal']
No_plates = [2,4,20,10,1,5]

plt.figure(figsize = (4,2))  # plt.figure for adjusting the size of the chart
plt.bar(breakfast_items,No_plates, color = ["violet", "green", "pink", "orange","blue", "gold"])  # plt.bar(x,y)

# plt.text(x,y,value) is used to add the text/value  in the chart
# plt.text(x,y, value) will place "value" at x and y corrdinate
# x : 0,1,2,3,4,5
# y : 2,4,20,10,1,5

# enumerate() is used to generate the index.
for x,y in enumerate(No_plates):
    plt.text(x,y+1,y)

plt.title("Food vs Plates")
plt.ylabel("No.of plates")
plt.xlabel("Food items")    
```

# Horizontal bar chart

```python
# Def: Categorical data is plotted on y axis
# Big item names are allowed in Horizontal bar chart.
# Horizontal bar chart can handle big categorical names without overlapping.

breakfast_items = ['Idli','Dosa','vada','Puri','Bonda','PaneerButtorMasalaWithCashewCreamAndTomatoFlavour']
No_plates = [2,4,20,10,1,5]

plt.barh(breakfast_items,No_plates)
plt.title("Food vs Plates")
plt.ylabel("Food items")
plt.xlabel("No.of plates") 

for x,y in enumerate(No_plates):
    plt.text(y,x,x)
```

# Histogram

```python
# USE: This chart is use for representing Continous data.
# Histogram is popularly used for representing the shape of the data.

import numpy as np
data = np.random.randn(1000)

plt.figure(figsize = (3,2))
plt.hist(data, bins = 30, edgecolor = 'black')
plt.title("Normal Distribution")
plt.xlabel("Continuous data")
plt.ylabel("Total value = 1000")
plt.show()
```

# Line chart

```python
# USE : Trends over the time

dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')
dates
```

```python
Icrecream_sales = [ 1000, 90, 80, 70, 250, 400, 450, 500, 10, 20, 50, 60]
```

```python
# plot() function is used to plot the line chart
plt.figure(figsize = (4,3))
plt.plot(dates,Icrecream_sales, marker = 'o') 


# xticks() function is used for tilt the values on x axis with specified angle
# '%b' : 3 character month name
# '%B' : full month name
# '%Y' : year
plt.xticks(dates, dates.strftime('%b'),rotation = 45)

plt.title("Icecream sales over the month")
plt.xlabel("Month names")
plt.ylabel("No.of Icecream sales") 

plt.grid(True)
plt.show()
```

# Multiline chart

```python
# USE : multiline chart is used for comparing multiple trends side by side

dates = pd.date_range(start = "2012-01-01",periods = 12, freq = 'YE')
dates
```

```python
pawan_kalyan_fan = [ 50,60,70,100,80,90,100,110,120,130,140,150]
prabhas_fan = [10,30,50,60,70,80,90,100,140,200,250,50]
rajinikant_fan = [5,20,50,100,150,200,300,350,400,450,500,550]

plt.plot(dates,pawan_kalyan_fan, marker = 'o', label = "Pawan")
plt.plot(dates,prabhas_fan, marker = '*', label = "Prabhas")
plt.plot(dates,rajinikant_fan, marker = '+',label = "rajinikant")

# legend() provide the detail of different lines
# legend() labels the different category of data
plt.legend()
```

# Scatter plot

```python
# USE : used to find the relation between two numerical variables

# Type of relation:
# 1. Positive relation (Both move in same direction)
# 2. Negative relation (Both move in opposite direction)
# 3. No relation 



Age = [ 16,18,21,30,40,20,35,32,23,25,22,19,26]
Salary = [20,25,30,40,50,22,45,42,27,29,26,27,20]

plt.figure(figsize = (3,2))
plt.scatter(Age, Salary)

plt.xlabel("Age")
plt.ylabel("Salary")
```

# Donut chart

```python
# Donut chart is a variation of pie chart.
# Donut chart is a pie chart with space at the center.
# The center space of donut chart  can be used for adding more information.
market_share_mobiles = ["OnePlus", "IQ", "Vivo", " Apple", "Samsumg", "Realme", "OPPO", "other"]
share = [ 5,1,15,20,25,19,10,5]


plt.figure(figsize = (4,3))
plt.pie(share, labels = market_share_mobiles, autopct = "%1.1f%%", wedgeprops = dict(width = 0.6))
plt.title("Mobile Market Share") 

plt.text(0,0, "Sales", ha = "center")
```

```python
import matplotlib.pyplot as plt
```

```python
Mandi = ["Biriyani", "Chicken","Mutton","Dry fruits","Vegitables","Kaju","Onion","Raita"]
share = [25,20,20,5,10,15,5,10,]

plt.subplot(2,3,1)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.subplot(2,3,2)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.subplot(2,3,3)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.subplot(2,3,4)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.subplot(2,3,5)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.subplot(2,3,6)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.tight_layout()
plt.show()
```

```python
Mandi = ["Biriyani", "Chicken","Mutton","Dry fruits","Vegitables","Kaju","Onion","Raita"]
share = [25,20,20,5,10,15,5,10]

plt.subplot(2,2,1)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.subplot(2,2,2)
plt.barh(Mandi,share)

plt.subplot(2,2,3)
plt.barh(Mandi,share)

plt.subplot(2,2,4)
plt.pie(share, labels = Mandi, autopct = "%1.1f%%")

plt.tight_layout()
plt.show()
```

# method2:- plt.subplots() : create all plots at once .it is good to plot large number of subplots

```python
gym_data = ["Trainer", "Selfie girls", "Helping_guy", "New_year_resolution", "Slient_lifter", "Content_creater"]
per = [5, 25, 10, 15, 25, 20]

# fig : the whole figure/canvas
# figsize : size of the complete canvas
# axes : array of subplots ( 2 X 3)
fig, axes = plt.subplots(2,2)

# ravel() covert 2D to 1D
one_d_array = axes.ravel()

labels = ['pie1','pie2','pie3','pie4']

for ax,label in zip(one_d_array,labels):
    ax.pie(per, labels = gym_data)
    ax.set_title(label)

# set_title() function is used to set the title of subplots
# zip() : to access the data element by element from multiple subarrays
    
plt.tight_layout()
plt.show()
```

# 3D chart using matplotlib

```python
# 3D charts can also be printed in matplotlib
# although plotly is much better for plotting 3D charts.
import matplotlib.pyplot as plt
import numpy as  np
```

```python
# plt.figure() is used to set the size of image frame
figure = plt.figure()

#add_subplot() have 3d projection parameter.
# which helps to plot 3d charts
# 111 : Row Column Index
ax = figure.add_subplot(111, projection = '3d')

# generating 100 random elements for each dimension/variable x,y,z
x, y, z = np.random.rand(3, 100)

# plotting the scatter plot for 3 variables/dimension
ax.scatter(x,y,z)
plt.show()
```
